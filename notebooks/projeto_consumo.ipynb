```python
## 1. Inicialização do Ambiente e Configurações
# Importação da biblioteca `pandas`, que atua como o motor fundamental para a manipulação, estruturação e análise de matrizes de dados em formato tabular dentro do ecossistema Python.

import pandas as pd
```

```python
## 2. Preparação dos Dados: Extração e Higienização Técnica (ETL)

# Esta etapa consolida o processo de transformação de dados brutos extraídos do ecossistema logístico em informações estruturadas e confiáveis para análise. O pipeline executa três procedimentos críticos de saneamento de forma linear:

# 1. **Extração com Tratamento de Metadados (`pd.read_excel`):** Aponta para o diretório de arquivos brutos (`raw`) e utiliza o parâmetro `header=3`. Essa decisão técnica descarta as linhas iniciais de lixo sistêmico e metadados gerados pelo ERP, alinhando o cabeçalho do DataFrame diretamente com a granularidade dos SKUs.
# 2. **Expurgo de Colunas Fantasmas (`df.columns.str.contains`):** Executa uma varredura via expressão regular para identificar e eliminar colunas nulas rotuladas automaticamente como `Unnamed`. Isso remove artefatos de formatação vazios vindos da exportação original, otimizando o consumo de memória.
# 3. **Supressão de Linhas Nulas (`df.dropna`):** Elimina registros completamente em branco do DataFrame (`how='all'`), garantindo que apenas linhas contendo interações operacionais e saldos reais persistam na base para as regras de negócio e cálculo dos KPIs logísticos.

caminho = "../data/raw/projeto_consumo.xlsx"
df = pd.read_excel(caminho, header=3)
```

```python

df = df.dropna(how='all')
```
