```python
import pandas as pd
```

```python
caminho = r"../data/raw/projeto_consumo.xlsx"
```

```python
df = pd.read_excel(caminho, header=3)
```

```python
df = df.loc[:, ~df.columns.str.contains('^Unnamed')]
```

```python
df = df.dropna(how='all')
```

```python
colunas_interesse = [
    'SKU', 'CATEGÓRIA', 'DESCRIÇÃO', 'UNID. MEDIDA', 
    'ESTOQUE ATUAL', 'CONSUMO MÉDIO MENSAL', 'CURVA ABC', 
    'ITEM ESSENCIAL', 'VALOR UNIT. (R$)', 'VALOR TOTAL (R$)','NF-e', 'FORNECEDOR'
]
```

```python
df = df[colunas_interesse]
```
